```python
import pandas as pd

file_path = "../data/Global_Landslide_Catalog_Export_20250201 (1).csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)
```

```python
print("Dataset Columns:")
for column in df.columns:
    print(column)
```

```python

df.head()
```

```python
useful_columns = [
    "event_date",
    "latitude",
    "longitude",
    "country_name",
    "admin_division_name",
    "landslide_category",
    "landslide_trigger",
    "landslide_size",
    "landslide_setting",
    "fatality_count",
    "injury_count"
]

df[useful_columns].head(10)
```

```python
print("Landslide Triggers:")
print(df["landslide_trigger"].value_counts().head(15))

print("\nLandslide Sizes:")
print(df["landslide_size"].value_counts())

print("\nLandslide Categories:")
print(df["landslide_category"].value_counts())
```

```python
print("Countries with most landslide records:")
print(df["country_name"].value_counts().head(20))
```

```python
print("\nIndia records:")
print((df["country_name"] == "India").sum())
```

```python
print("\nIndian states/regions in the dataset:")
india_df = df[df["country_name"] == "India"]

print(india_df["admin_division_name"].value_counts().head(30))
```

```python
ner_states = [
    "Assam",
    "Nāgāland",
    "Nagaland",
    "Manipur",
    "Mizoram",
    "Sikkim",
    "Arunāchal Pradesh",
    "Arunachal Pradesh",
    "Meghalaya",
    "Meghālaya",
    "Tripura"
]

ner_df = df[
    (df["country_name"] == "India") &
    (df["admin_division_name"].isin(ner_states))
].copy()

print("NER records:", len(ner_df))
print("\nNER states:")
print(ner_df["admin_division_name"].value_counts())
```

```python
state_name_mapping = {
    "Nāgāland": "Nagaland",
    "Nagaland": "Nagaland",
    "Arunāchal Pradesh": "Arunachal Pradesh",
    "Arunachal Pradesh": "Arunachal Pradesh",
    "Meghālaya": "Meghalaya",
    "Meghalaya": "Meghalaya"
}

ner_df["admin_division_name"] = ner_df["admin_division_name"].replace(
    state_name_mapping
)

print("Cleaned NER state distribution:")
print(ner_df["admin_division_name"].value_counts())
```

```python
# Keep only records with known landslide sizes
model_df = ner_df[
    ner_df["landslide_size"].isin(["small", "medium", "large"])
].copy()

# Convert landslide size into risk levels
risk_mapping = {
    "small": "LOW",
    "medium": "MEDIUM",
    "large": "HIGH"
}

model_df["risk_level"] = model_df["landslide_size"].map(risk_mapping)

print("Total usable records:", len(model_df))

print("\nRisk Level Distribution:")
print(model_df["risk_level"].value_counts())
```

```python
feature_columns = [
    "latitude",
    "longitude",
    "admin_division_name",
    "landslide_trigger",
    "landslide_category",
    "landslide_setting",
    "fatality_count",
    "injury_count",
    "event_date"
]

print(model_df[feature_columns].isnull().sum())
```

```python
# Convert event_date to datetime
model_df["event_date"] = pd.to_datetime(
    model_df["event_date"],
    errors="coerce"
)

# Extract year and month
model_df["event_year"] = model_df["event_date"].dt.year
model_df["event_month"] = model_df["event_date"].dt.month

print("Missing dates after conversion:")
print(model_df["event_date"].isnull().sum())

print("\nYear range:")
print(model_df["event_year"].min(), "to", model_df["event_year"].max())

print("\nMonth distribution:")
print(model_df["event_month"].value_counts().sort_index())
```

```python
# Final input features
feature_columns = [
    "latitude",
    "longitude",
    "event_month",
    "admin_division_name",
    "landslide_trigger",
    "landslide_category",
    "landslide_setting"
]

# Input features
X = model_df[feature_columns]

# Target
y = model_df["risk_level"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data shape:", X_train.shape)
print("Testing data shape:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

# Separate numerical and categorical columns
numerical_features = [
    "latitude",
    "longitude",
    "event_month"
]

categorical_features = [
    "admin_division_name",
    "landslide_trigger",
    "landslide_category",
    "landslide_setting"
]

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        ("num", "passthrough", numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Preprocessor created successfully!")
print("Numerical features:", numerical_features)
print("Categorical features:", categorical_features)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

# Create the Random Forest model
model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

# Combine preprocessing + model
pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", model)
])

# Train the model
pipeline.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Make predictions on unseen test data
y_pred = pipeline.predict(X_test)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", round(accuracy * 100, 2), "%")

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))
```

```python
balanced_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight={
        "LOW": 2,
        "MEDIUM": 1,
        "HIGH": 4
    }
)

balanced_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", balanced_model)
])

balanced_pipeline.fit(X_train, y_train)

balanced_pred = balanced_pipeline.predict(X_test)

print("Balanced Random Forest trained successfully!")
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Predictions
balanced_pred = balanced_pipeline.predict(X_test)

# Accuracy
balanced_accuracy = accuracy_score(y_test, balanced_pred)

print("Balanced Model Accuracy:",
      round(balanced_accuracy * 100, 2), "%")

print("\nBalanced Classification Report:")
print(classification_report(y_test, balanced_pred))

print("\nBalanced Confusion Matrix:")
print(confusion_matrix(y_test, balanced_pred))
```

```python
# Create binary high-risk target
model_df["high_risk"] = (
    model_df["risk_level"] == "HIGH"
).astype(int)

print("High-risk target distribution:")
print(model_df["high_risk"].value_counts())

print("\nHigh-risk percentage:")
print(
    round(model_df["high_risk"].mean() * 100, 2),
    "%"
)
```

```python
# Input features
X_binary = model_df[feature_columns].copy()

# Binary target
y_binary = model_df["high_risk"].copy()

print("X_binary shape:", X_binary.shape)
print("y_binary shape:", y_binary.shape)

print("\nTarget distribution:")
print(y_binary.value_counts())
```

```python

from sklearn.model_selection import train_test_split

X_train_bin, X_test_bin, y_train_bin, y_test_bin = train_test_split(
    X_binary,
    y_binary,
    test_size=0.20,
    random_state=42,
    stratify=y_binary
)

print("Training data shape:", X_train_bin.shape)
print("Testing data shape:", X_test_bin.shape)

print("\nTraining target distribution:")
print(y_train_bin.value_counts())

print("\nTesting target distribution:")
print(y_test_bin.value_counts())
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline

high_risk_model = RandomForestClassifier(
    n_estimators=500,
    random_state=42,
    class_weight="balanced",
    min_samples_leaf=2
)

high_risk_pipeline = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", high_risk_model)
])

high_risk_pipeline.fit(X_train_bin, y_train_bin)

print("High-risk detection model trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# Predict on unseen test data
y_pred_bin = high_risk_pipeline.predict(X_test_bin)

# Accuracy
accuracy = accuracy_score(y_test_bin, y_pred_bin)

print("High-Risk Model Accuracy:",
      round(accuracy * 100, 2), "%")

# Classification report
print("\nClassification Report:")
print(
    classification_report(
        y_test_bin,
        y_pred_bin,
        target_names=["Normal", "High Risk"],
        zero_division=0
    )
)

# Confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test_bin, y_pred_bin))
```

```python
# Get probability of High Risk
y_prob = high_risk_pipeline.predict_proba(X_test_bin)[:, 1]

print("High-risk probabilities:")
print(y_prob)
```

```python
import pandas as pd

probability_results = pd.DataFrame({
    "Actual": y_test_bin.values,
    "High_Risk_Probability": y_prob
})

print(probability_results.sort_values(
    "High_Risk_Probability",
    ascending=False
).head(15))
```

```python
def predict_risk(input_data):
    """
    Predict landslide high-risk probability
    and convert it into a dashboard risk level.
    """

    probability = high_risk_pipeline.predict_proba(input_data)[0][1]

    if probability >= 0.70:
        risk_level = "HIGH"
    elif probability >= 0.40:
        risk_level = "MODERATE"
    else:
        risk_level = "LOW"

    return {
        "high_risk_probability": round(float(probability), 4),
        "risk_percentage": round(float(probability) * 100, 2),
        "risk_level": risk_level
    }
```

```python
sample_input = X_test_bin.iloc[[0]]

result = predict_risk(sample_input)

print(result)
```
